```python
plt.figure(figsize=(7,5))

sns.countplot(x="LeftCompany", data=df)

plt.title("Employee Attrition Distribution")
plt.xlabel("Left Company")
plt.ylabel("Number of Employees")

plt.show()
```

```python
df = df.drop(["EmployeeID", "FullName"], axis=1)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], kde=True)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["MonthlySalary"], kde=True)

plt.title("Monthly Salary Distribution")
plt.xlabel("Monthly Salary")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.countplot(
    x="Department",
    hue="LeftCompany",
    data=df
)

plt.title("Employee Attrition by Department")
plt.xticks(rotation=30)

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    x="DoesOvertime",
    hue="LeftCompany",
    data=df
)

plt.title("Employee Attrition by Overtime")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="WorkLifeBalance",
    hue="LeftCompany",
    data=df
)

plt.title("Employee Attrition by Work-Life Balance")

plt.show()
```

```python
numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

numeric_columns
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df[numeric_columns].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```
